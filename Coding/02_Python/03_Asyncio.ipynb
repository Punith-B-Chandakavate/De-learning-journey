# 1. Basic Asyncio

```python
import asyncio
import time

async def worker(name):
    print(f"Task {name} started")

    await asyncio.sleep(2)

    print(f"Task {name} finished")

async def main():
    await worker("1")
    await worker("2")

asyncio.run(main())
```

# 2. Run tasks concurrently with asyncio.gather()

```python
import asyncio
import time

async def worker(name):
    print(f"Task {name} started")

    await asyncio.sleep(2)

    print(f"Task {name} finished")

async def main():
    await asyncio.gather(
        worker("1"),
        worker("2"),
        worker("3")
    )
```

# 3. Asyncio with API calls

```python
import asyncio

async def fetch_data(api_name):
    print(f"Calling {api_name}")

    await asyncio.sleep(2)

    print(f"{api_name} response received")

    return f"{api_name} data"

async def main():
    results = await asyncio.gather(
        fetch_data("Customer Api"),
        fetch_data("Order API"),
        fetch_data("Product API")
    )

    print("\n Result:")

    for result in results:
        print(result)

asyncio.run(main())
```

# 4. Using asyncio.create_task()

```python
import asyncio

async def worker(name):
    print(f"{name} started")
    await asyncio.sleep(2)
    print(f"{name} finished")
    return f"{name} result"

async def main():
    t1 = asyncio.create_task(worker("task 1"))
    t2 = asyncio.create_task(worker("task 2"))
    t3 = asyncio.create_task(worker("task 3"))

    res1 = await t1
    res2 = await t2
    res3 = await t3

    print(res1)
    print(res2)
    print(res3)

asyncio.run(main())
```

# 5. Practical Data Engineering example

```python
import asyncio
import random


async def fetch_api(api_name):

    print(f"Fetching {api_name}")

    delay = random.uniform(1, 3)

    await asyncio.sleep(delay)

    return {
        "api": api_name,
        "status": "success"
    }


async def main():

    apis = [
        "Customer API",
        "Order API",
        "Product API",
        "Payment API",
        "Inventory API"
    ]

    results = await asyncio.gather(
        *(fetch_api(api) for api in apis)
    )

    for result in results:
        print(result)


asyncio.run(main())
```
