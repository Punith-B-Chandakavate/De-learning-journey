# 1. Basic multiprocessing

```python
from multiprocessing import Process

def task(name):
    print(f"{name} started")
    time.sleep(2)
    print(f"{name} completed")


if __name__ == "__main__":

    p1 = Process(target=task, args=("Process 1",))
    p2 = Process(target=task, args=("Process 2",))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print("All process completed")
```

# 2. Multiprocessing with CPU-heavy calculation

```python
from multiprocessing import Process
import time

def calculate(start, end):
    total = 0

    for i in range(start, end):
        total += i * i

    print(f"Calculation completed: {start} - {end}")
    print(f"Result: {total}")


if __name__ == "__main__":

    start_time = time.time()

    p1 = Process(target=calculate, args=(0, 10_000_000))
    p2 = Process(target=calculate, args=(10_000_000, 20_000_000))

    p1.start()
    p2.start()

    p1.join()
    p2.join()

    print(f"Time taken: {time.time() - start_time:.2f} seconds")
```

# 3. Multiprocessing Pool

```python
from multiprocessing import Pool

def square(number):
    return number * number

if __name__ == "__main__":

    number = [1,2,3,4,5,6,7,8]
    with Pool(processes=4) as pool:
        results = pool.map(square, number)

    print(results)
```

# 4. Multiprocessing + Data Engineering example

```python
from multiprocessing import Pool
import time

def process_file(file):
    print(f"Processing {file}")

    total = 0
    for i in range(5000000):
        total += i * i
    return f"{file} processed"

if __name__ == "__main__":
    files = [
        "file1.csv",
        "file2.csv",
        "file3.csv",
        "file4.csv"
    ]

    with Pool(process=4) as pool:
        results = pool.map(process_file, files)

    for result in results:
        print(result)
    
```
